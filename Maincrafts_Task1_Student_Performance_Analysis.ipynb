```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
url = "https://raw.githubusercontent.com/MakroMB/student-performance-analysis/main/data/student-mat.csv"

df = pd.read_csv(url, sep=";")

print("Dataset loaded successfully!")
```

## Data Cleaning

Data cleaning is performed to ensure that the dataset is accurate and consistent.

The following operations were performed:

1. Checked missing values using `isnull()`.
2. Checked duplicate records using `duplicated()`.
3. Removed duplicate records using `drop_duplicates()`.
4. Verified the dataset after cleaning.

### Data Cleaning Results
- Missing Values: 0
- Duplicate Records: 0
- Total Records: 395
- Total Columns: 33

The dataset was already clean, so no records were removed.

```python
df.isnull().sum()
```

```python
print("Total Missing Values:", df.isnull().sum().sum())
```

```python
print("Duplicate Records:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Dataset after removing duplicates:", df.shape)
```

```python
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Records:", df.duplicated().sum())
```

```python
average_grade = df["G3"].mean()

print("Average Final Grade:", round(average_grade, 2))
```

```python
students_above_15 = (df["G3"] > 15).sum()

print("Students scoring above 15:", students_above_15)
```

```python
correlation = df["studytime"].corr(df["G3"])

print("Correlation:", round(correlation, 3))
```

```python
gender_performance = df.groupby("sex")["G3"].mean()

print(gender_performance)
```

```python
if gender_performance["F"] > gender_performance["M"]:
    print("Female students perform better on average.")
else:
    print("Male students perform better on average.")
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["G3"], bins=10, edgecolor="black")

plt.title("Distribution of Students' Final Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(x="studytime", y="G3", data=df)

plt.title("Study Time vs Final Grades")
plt.xlabel("Study Time")
plt.ylabel("Final Grade (G3)")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.barplot(x="sex", y="G3", data=df)

plt.title("Average Final Grade by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Final Grade")

plt.xticks([0, 1], ["Female", "Male"])

plt.show()
```

## Visualization Insights

1. **Grade Distribution:** The histogram shows how students' final grades are distributed across different score ranges.

2. **Study Time vs Final Grades:** The scatter plot shows a very weak positive relationship between study time and final grades.

3. **Gender-wise Performance:** The bar chart shows that male students have a slightly higher average final grade (10.91) than female students (9.97).
